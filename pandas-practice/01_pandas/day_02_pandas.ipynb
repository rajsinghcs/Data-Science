# Data Cleaning

```python
df.isnull().sum()
```

```python
df
```

```python
df.dropna(how = "any") # if any value in any row found then remove
# df.dropna(how = "all") # if all values are null in any row then remove
```

```python
#Fill missing values
df["Ages"].fillna(df["Ages"].mean()) #Mean
# df["Ages"].fillna(df["Ages"].median()) #Media
# df["Ages"].fillna(df["Ages"].mode()) #Mode
```

```python
df["Ages"].fillna(method = "ffill")  #ForwardFill
```

```python
df["Ages"].fillna(method = "bfill")  #BackwardFill
#still NaN beacuse there is not value after that nan value so it stay NaN
```

```python
df["Name"] = df["Names"].replace("Priya","Riya")
df
```

```python
#Duplicate Values
df.duplicated(keep ="first")
```

```python
df = df.drop_duplicates()
df
```

```python
#Invalid values
#lambda
df["Promoted Employees"] = df["Promoted Employees"].apply(lambda x : x/10 if x>55000 else x)
df
```

```python
#apply nad lambda
def multiplay_age(x):
    return x*2
df["Ages" ]= df["Ages"].apply(multiplay_age)
df
```

```python
df["Ages"] = df["Ages"].apply(lambda x: x/2)
df
```
