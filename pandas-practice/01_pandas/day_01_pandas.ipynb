```python
import pandas as pd #alias
```

```python
s = pd.Series([20,30,40,10,50])
s
```

```python
s.name = "Calories"
```

# 1.Inspection / Information Methods

```python
print(s.head())        # First 5 values
print(s.tail())        # Last 5 values
print(s.info())        # Summary info
print(s.describe())    # Statistical summary
print(s.dtype)         # Data type
print(s.shape)         # Size of Series
print(s.size)          # Number of elements
print(s.index)         # Index labels
print(s.values)        # Underlying NumPy array
```

# 2. Statistical / Aggregation Methods

```python
print(s.sum())      # Sum of values
print(s.mean())     # Average
print(s.median())   # Median
print(s.min())      # Minimum
print(s.max())      # Maximum
print(s.std())      # Standard deviation
print(s.var())      # Variance
print(s.count())    # Non-null values
```

# 3. Sorting Methods

```python
print(s.sort_values())   # Sort values
print(s.sort_index())    # Sort by index
```

# 4. Indexing / Selection Methods

```python
# Methods based indexing
print(s.loc[1])   # Access by label
print(s.iloc[1])  # Access by position
print(s.at[1])    # Fast label access
print(s.iat[1])   # Fast integer access
```

```python
#Indexing
print(s[0])
```

```python
print(s[1:3])
```

```python
s[2:4]
```

# 5. Value Counting & Unique

```python
print(s.value_counts())   # Frequency
print(s.unique())         # Unique values
print(s.nunique())        # Count unique
```

# 6. Transformation Methods

```python
print(s.apply(lambda x: x*2))   # Apply function
print(s.map(lambda x: x+5))     # Element mapping
print(s.astype(float))          # Change datatype
```

# 7. Missing Value Handling

```python
print(s.isna())      # Detect NaN
print(s.notna())     # Non-null check
print(s.dropna())    # Remove missing
print(s.fillna(0))   # Replace NaN
```

#  8. Mathematical Operations

```python
print(s + 10)  #add
print(s * 2)   #multiply
print(s ** 2)  #power
```

# 9. Conversion Methods

```python
print(s.to_list())    # Convert to list
print(s.to_numpy())   # Convert to NumPy
print(s.to_dict())    # Convert to dictionary
```

```python
fruits_protein = {
    "Apple": 0.3,
    "Banana": 1.3,
    "Orange": 0.9,
    "Guava": 2.6,
    "Mango": 0.8,
    "Papaya": 0.5,
    "Pomegranate": 1.7,
    "Strawberry": 0.7
}
```

```python
s2 = pd.Series(fruits_protein)
s2
```

```python
s2.name = "Protein"
```

```python
s2.name
```

## Conditional Selection

```python
s2>1  #protein greater than 1 -> True
```

```python
#now i want to print only those fruits which have protein greater than 1
s2[s2>1]
```

## Logical Operator

```python
#And , Or , Not
#AND
s2[(s2 > 0.5) & (s2 < 1)]
```

```python
#OR
s2[(s2 == 0.7 ) | (s2 == 0.9)]
```

```python
#NOT
s2[~(s2 == 0.7)]
```
