```python
import pandas as pd
data = {
    "Age": [21, 25, 30, 22, 28],
    "Salary": [25000, 40000, 50000, 27000, 45000],
    "Experience": [1, 3, 5, 2, 4],
    "Dept":["HR","IT","Marketing","Sales","HR"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Multivariate Analysis: 3 numerical Column
#Bubble Plot
import matplotlib.pyplot as plt
plt.title("Age vs Salary vs Experience")
plt.scatter(df["Age"],df["Salary"],s=df["Experience"]*80, color= "red", edgecolors="black")
```

```python
# 2 Numerical column and 1 categorial column
color={"HR":"black","Sales":"red","Marketing":"orange","IT":"blue"}
for Dept, color in color.items():
    df_dept = df[df["Dept"] == Dept]
    plt.scatter(df_dept["Age"], df_dept["Salary"], c=color, label = Dept, s=150)
plt.legend()

plt.title("Age vs Salary by Department")
```

```python
sort_ages = df.sort_values(by="Age")
sort_ages
```

```python
#Object Oriented API
fig, axs = plt.subplots(1, 2, figsize=(10,5))

#Line Plot
axs[0].plot(sort_ages["Age"],df["Salary"], color="red", marker="*",linewidth= "2", markersize="2")
axs[0].grid()
axs[0].set_title("Line Plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")


axs[1].boxplot(sort_ages["Salary"])
axs[1].set_title("Boxplot of Salary")
axs[1].set_ylabel("Salary")
axs[1].grid()

plt.savefig("multipleplots.png")

plt.show()
```
