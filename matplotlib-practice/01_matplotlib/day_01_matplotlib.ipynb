# Pyplot API

```python
#Univariate - Numerical
data = {
    "Salary":[2000,3000,45000,5000,60030,7300,80030,90300,10000]
}
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
#Line Plot
plt.plot(df,color = 'red',marker = "o")
plt.grid()
#linestyle, linewidth
```

```python
#Histogram
plt.hist(df,bins = 3, color = "red")
```

```python
#Box Plot
plt.boxplot(df)  #mo outlier in this data
```

```python
#adding outliers
df.loc[10] = [300000]
plt.boxplot(df)
```

```python
df.drop(index = 10, inplace = True) #delete outliers
df
```

```python
#Univariate Categorial value
df["Dept"] = ['HR','IT','Finance','HR','IT','Finance','HR','IT','Finance']
df.head()
```

```python
#Pie Chart - it is used for categorial value
count = df["Dept"].value_counts()
count
```

```python
plt.pie(count, labels = count.index, autopct="%1.1f", explode=[0,0.1,0.1])
```

```python
#Countplot
plt.bar(count.index, count, color = ['green','blue','red'])
```

```python
#Bivariate Analysis
df.head()
```

```python
df["Ages"] = [44,53,22,25,34,54,50,49,59]
```

```python
df
```

```python
#Scatter Plot
plt.scatter(df["Ages"], df["Salary"],color = 'red')
```

```python
#Line plot
plt.plot(df["Ages"],df["Salary"], color='red', marker='o', linewidth=2)
```

```python
#Bar chart
plt.bar(df["Ages"], df["Salary"], color= 'blue')
```

```python
#Bivariate Numerical and categorial features
hr_salary = df[df["Dept"] == "HR"]["Salary"]
```

```python
hr_salary
```

```python
it_salary = df[df["Dept"] == "IT"]["Salary"]
finance_salary = df[df["Dept"] == "Finance"]["Salary"]
```

```python
#BoxPlot
plt.boxplot([hr_salary,it_salary,finance_salary], labels=['HR', 'IT','Finance']);
```

```python
#Pie Chart
salry_by_dept = df.groupby("Dept")["Salary"].sum()
salry_by_dept
```

```python
plt.pie(salry_by_dept,labels=salry_by_dept.index , autopct="%1.1f", explode=[0.1
                                                                             ,0.1,0.1])
```

```python
#Bar Plot
hr_mean = sum(hr_salary)/len(hr_salary)
it_mean = sum(it_salary)/len(it_salary)
finance_mean = sum(finance_salary)/len(finance_salary)
```

```python
plt.bar(['HR','IT','Finance'], [hr_mean,it_mean,finance_mean], color=["red","green","blue"])
plt.grid()
```
