## 1.Numpy Arrays

```python
range = np.arange(1,10,2) # start stop and step
range
```

### linspace evenly space number

```python
arr = np.linspace(0,1,10)
arr
```

# logSpace

```python
log_arr = np.logspace(1,3,3)
log_arr
```

### array of zeros

```python
arr_zero = np.zeros(5)
arr_zero
```

```python
arr_zero_2d = np.zeros([4,3])
arr_zero_2d
```

### array of one

```python
arr_one = np.ones([2,4], dtype=int)
arr_one
```

### array of any values

```python
arr_value = np.full(10,2)
arr_value
```

```python
arr_value = np.full([2,2],2)  # row , col , value
arr_value
```

### uninitialized array

```python
arr_uni = np.empty([2,3])
arr_uni
```

### random float values

```python
arr = np.random.rand(5)
arr
```

### random randint

```python
arr = np.random.randint(10,100, size = (3,3))  # start ,stop , size
arr
```

## 2.Numpy Data Types and Type Casting

```python
arr = np.array([2,3,4,2,3,4])
arr
```

```python
arr = np.array([2,3,4,2.2,3,4])
print(type(arr))
```

```python
lst = ["string",3,2,1]
print(type(lst))
arr = np.array(lst)
print(type(arr))
```

```python
arr = np.array([4,3,2,4])
arr.dtype
```

### int to float and viceversa

```python
arr = np.array([4,3,2,4],dtype=float)
arr.dtype
```

```python
arr = np.array([1,2,3])
new_arr = arr.astype(np.float64)
new_arr.dtype
```

### Type Casting Errors

```python
arr = np.array(["1","2","Raj"])  #Invalid Literal Error
new_arr = arr.astype(np.int64)
new_arr.dtype
```

```python
a = np.array([200], dtype=np.int8)  #Overflow Error
```

```python
a = np.array([1.9999])  #Precision Loss
b = a.astype(int)

print(b)  # 1
```

```python
a = np.array([1+2j])  #Complex to Float Error
a.astype(float)
```
